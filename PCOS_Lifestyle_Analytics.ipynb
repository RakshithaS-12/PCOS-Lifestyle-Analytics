```python
# Define features and target

# Remove identifier and target columns from the input features
X = df.drop(
    columns=[
        "Participant_ID",
        "Overall_Lifestyle_Profile",
        "Primary_Recommendation_Area"
    ]
)

# Target variable
y = df["Overall_Lifestyle_Profile"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeature columns:")
print(X.columns.tolist())

print("\nTarget classes:")
print(y.value_counts())
```

```python
# Check all questionnaire response categories

response_columns = [
    "Period_Regularity_Response",
    "Cycle_Length_Response",
    "Missed_Periods_Response",
    "Menstrual_Flow_Response",
    "Period_Duration_Response",
    "Period_Symptoms_Response",
    "Acne_Response",
    "Excess_Hair_Response",
    "Hair_Loss_Response",
    "Diet_Response",
    "Exercise_Response",
    "Sleep_Response",
    "Stress_Response",
    "Water_Intake_Response",
    "Fast_Food_Response",
    "Daily_Activity_Response",
    "Recent_Weight_Gain_Response"
]

for column in response_columns:
    print("\n" + "=" * 70)
    print(column)
    print("=" * 70)
    for i, value in enumerate(df[column].unique(), start=1):
        print(i, "-", value)
```

```python
# Import preprocessing tools

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Numeric features
numeric_features = [
    "Age",
    "BMI"
]

# Categorical features
categorical_features = [
    "Period_Regularity_Response",
    "Cycle_Length_Response",
    "Missed_Periods_Response",
    "Menstrual_Flow_Response",
    "Period_Duration_Response",
    "Period_Symptoms_Response",
    "Acne_Response",
    "Excess_Hair_Response",
    "Hair_Loss_Response",
    "Diet_Response",
    "Exercise_Response",
    "Sleep_Response",
    "Stress_Response",
    "Water_Intake_Response",
    "Fast_Food_Response",
    "Daily_Activity_Response",
    "Recent_Weight_Gain_Response"
]

print("Numeric features:", len(numeric_features))
print("Categorical features:", len(categorical_features))
```

```python
# Create preprocessing pipeline

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            StandardScaler(),
            numeric_features
        ),
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ]
)

print("Preprocessing pipeline created successfully!")
```

```python
# Split data into training and testing sets

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
# Fit preprocessing pipeline on training data

X_train_processed = preprocessor.fit_transform(X_train)

# Transform testing data using the already-fitted preprocessor
X_test_processed = preprocessor.transform(X_test)

print("Training data after preprocessing:", X_train_processed.shape)
print("Testing data after preprocessing:", X_test_processed.shape)
```

```python
# Train Logistic Regression model

from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

logistic_model.fit(X_train_processed, y_train)

print("Logistic Regression model trained successfully!")
```

```python
# Make predictions on the test data

y_pred = logistic_model.predict(X_test_processed)

print("Predictions generated successfully!")
print("\nFirst 10 predictions:")
print(y_pred[:10])
```

```python
# Evaluate Logistic Regression model

from sklearn.metrics import accuracy_score, classification_report

accuracy = accuracy_score(y_test, y_pred)

print("Logistic Regression Accuracy:", round(accuracy, 4))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred
    )
)
```

```python
# Confusion Matrix

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=logistic_model.classes_
)

fig, ax = plt.subplots(figsize=(8, 6))

disp.plot(
    ax=ax,
    cmap="Blues",
    values_format="d"
)

plt.title("Logistic Regression Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")

plt.tight_layout()
plt.show()
```

```python
# Train Decision Tree model

from sklearn.tree import DecisionTreeClassifier

decision_tree_model = DecisionTreeClassifier(
    max_depth=6,
    random_state=42
)

decision_tree_model.fit(
    X_train_processed,
    y_train
)

print("Decision Tree model trained successfully!")
```

```python
# Make Decision Tree predictions

y_pred_dt = decision_tree_model.predict(X_test_processed)

print("Decision Tree predictions generated successfully!")

# Calculate accuracy
dt_accuracy = accuracy_score(y_test, y_pred_dt)

print("\nDecision Tree Accuracy:", round(dt_accuracy, 4))

# Classification report
print("\nDecision Tree Classification Report:")
print(
    classification_report(
        y_test,
        y_pred_dt
    )
)
```

```python
# Train Random Forest model

from sklearn.ensemble import RandomForestClassifier

random_forest_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=8,
    random_state=42
)

random_forest_model.fit(
    X_train_processed,
    y_train
)

print("Random Forest model trained successfully!")
```

```python
# Make Random Forest predictions

y_pred_rf = random_forest_model.predict(X_test_processed)

print("Random Forest predictions generated successfully!")

# Calculate accuracy
rf_accuracy = accuracy_score(y_test, y_pred_rf)

print("\nRandom Forest Accuracy:", round(rf_accuracy, 4))

# Classification report
print("\nRandom Forest Classification Report:")
print(
    classification_report(
        y_test,
        y_pred_rf
    )
)
```

```python
# Compare all three ML models

from sklearn.metrics import precision_score, recall_score, f1_score
import pandas as pd

model_comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ],
    "Accuracy": [
        accuracy_score(y_test, y_pred),
        accuracy_score(y_test, y_pred_dt),
        accuracy_score(y_test, y_pred_rf)
    ],
    "Macro Precision": [
        precision_score(y_test, y_pred, average="macro", zero_division=0),
        precision_score(y_test, y_pred_dt, average="macro", zero_division=0),
        precision_score(y_test, y_pred_rf, average="macro", zero_division=0)
    ],
    "Macro Recall": [
        recall_score(y_test, y_pred, average="macro", zero_division=0),
        recall_score(y_test, y_pred_dt, average="macro", zero_division=0),
        recall_score(y_test, y_pred_rf, average="macro", zero_division=0)
    ],
    "Macro F1": [
        f1_score(y_test, y_pred, average="macro", zero_division=0),
        f1_score(y_test, y_pred_dt, average="macro", zero_division=0),
        f1_score(y_test, y_pred_rf, average="macro", zero_division=0)
    ]
})

display(model_comparison.round(4))
```

```python
import shap

print("SHAP version:", shap.__version__)
print("SHAP imported successfully!")
```

```python
# Create SHAP explainer for Logistic Regression

explainer = shap.Explainer(
    logistic_model,
    X_train_processed,
    feature_names=feature_names
)

print("SHAP explainer created successfully!")
```

```python
# Calculate SHAP values for the test data

shap_values = explainer(X_test_processed)

print("SHAP values calculated successfully!")
print("SHAP values shape:", shap_values.values.shape)
```

```python
# Calculate overall SHAP feature importance

import numpy as np

# Take the absolute SHAP values
# Then average across test participants and the 3 classes
mean_shap = np.abs(shap_values.values).mean(axis=(0, 2))

# Create a feature-importance table
shap_importance = pd.DataFrame({
    "Feature": feature_names,
    "Mean_Absolute_SHAP": mean_shap
})

# Sort from most important to least important
shap_importance = shap_importance.sort_values(
    by="Mean_Absolute_SHAP",
    ascending=False
)

print("Top 15 features influencing the model:")

display(
    shap_importance.head(15).round(4)
)
```

```python
# Clean Top 15 SHAP Feature Importance Chart

top_features = shap_importance.head(15).copy()

# Create shorter labels
short_labels = []

for feature in top_features["Feature"]:
    if "Period_Regularity_Response" in feature:
        label = "Period Regularity"
    elif "Sleep_Response" in feature:
        label = "Sleep"
    elif "Diet_Response" in feature:
        label = "Diet"
    elif "Exercise_Response" in feature:
        label = "Exercise"
    elif "Stress_Response" in feature:
        label = "Stress"
    elif "Water_Intake_Response" in feature:
        label = "Water Intake"
    elif "Fast_Food_Response" in feature:
        label = "Fast Food"
    elif "Daily_Activity_Response" in feature:
        label = "Daily Activity"
    elif "Recent_Weight_Gain_Response" in feature:
        label = "Recent Weight Gain"
    else:
        label = feature

    short_labels.append(label)

top_features["Short_Feature"] = short_labels

# Sort for horizontal bar chart
top_features = top_features.sort_values(
    by="Mean_Absolute_SHAP",
    ascending=True
)

plt.figure(figsize=(10, 7))

bars = plt.barh(
    top_features["Short_Feature"],
    top_features["Mean_Absolute_SHAP"]
)

plt.title(
    "Top 15 Features Influencing Lifestyle Profile Predictions",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Mean Absolute SHAP Value", fontsize=11)
plt.ylabel("Lifestyle Factor", fontsize=11)

# Add SHAP values at the end of each bar
for bar in bars:
    value = bar.get_width()
    plt.text(
        value + 0.01,
        bar.get_y() + bar.get_height()/2,
        f"{value:.2f}",
        va="center",
        fontsize=9
    )

plt.tight_layout()
plt.show()
```

```python
# Aggregate SHAP importance by original lifestyle factor

shap_grouped = shap_importance.copy()

def get_original_feature(feature):
    for col in numeric_features + categorical_features:
        if col in feature:
            return col
    return feature

shap_grouped["Original_Feature"] = shap_grouped["Feature"].apply(
    get_original_feature
)

# Sum SHAP importance of all encoded categories
grouped_importance = (
    shap_grouped
    .groupby("Original_Feature")["Mean_Absolute_SHAP"]
    .sum()
    .sort_values(ascending=False)
)

# Display the grouped importance
grouped_importance
```

```python
# Plot aggregated SHAP importance

top_grouped = grouped_importance.head(10).sort_values(
    ascending=True
)

# Short display names
display_names = {
    "Period_Regularity_Response": "Period Regularity",
    "Cycle_Length_Response": "Cycle Length",
    "Missed_Periods_Response": "Missed Periods",
    "Menstrual_Flow_Response": "Menstrual Flow",
    "Menstrual_Duration_Response": "Period Duration",
    "Period_Duration_Response": "Period Duration",
    "Period_Symptoms_Response": "Period Symptoms",
    "Acne_Response": "Acne",
    "Excess_Hair_Response": "Excess Hair",
    "Hair_Loss_Response": "Hair Loss",
    "Diet_Response": "Diet",
    "Exercise_Response": "Exercise",
    "Sleep_Response": "Sleep",
    "Stress_Response": "Stress",
    "Water_Intake_Response": "Water Intake",
    "Fast_Food_Response": "Fast Food",
    "Daily_Activity_Response": "Daily Activity",
    "Recent_Weight_Gain_Response": "Recent Weight Gain",
    "Age": "Age",
    "BMI": "BMI"
}

labels = [
    display_names.get(feature, feature)
    for feature in top_grouped.index
]

plt.figure(figsize=(10, 6))

bars = plt.barh(
    labels,
    top_grouped.values
)

plt.title(
    "Lifestyle Factors Influencing Lifestyle Profile Predictions",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Aggregated Mean Absolute SHAP Value")
plt.ylabel("Lifestyle Factor")

# Add values
for bar, value in zip(bars, top_grouped.values):
    plt.text(
        value + 0.01,
        bar.get_y() + bar.get_height() / 2,
        f"{value:.2f}",
        va="center",
        fontsize=9
    )

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("PCOS_Lifestyle_Data.csv")

print("Dataset shape:", df.shape)
display(df.head())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression

# Target variable
target = "Overall_Lifestyle_Profile"

# Columns to exclude
X = df.drop(
    columns=["Participant_ID", "Overall_Lifestyle_Profile",
             "Primary_Recommendation_Area"]
)

y = df[target]

# Identify numerical and categorical features
numeric_features = ["Age", "BMI"]

categorical_features = [
    col for col in X.columns
    if col not in numeric_features
]

print("Number of features:", len(X.columns))
print("Numerical features:", numeric_features)
print("Categorical features:", len(categorical_features))
```

```python
# Split the dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
# Create preprocessing pipeline

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", StandardScaler(), numeric_features),
        ("categorical", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

# Fit on training data and transform both sets
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Processed training data:", X_train_processed.shape)
print("Processed testing data:", X_test_processed.shape)
```

```python
from sklearn.linear_model import LogisticRegression

# Train Logistic Regression model
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

logistic_model.fit(
    X_train_processed,
    y_train
)

# Predict on test data
y_pred = logistic_model.predict(X_test_processed)

print("Logistic Regression model trained successfully.")
print("Test predictions generated:", len(y_pred))
```
